# 1. Chapter objective

By the end of this chapter, you should be able to explain MLflow clearly from three angles:

1. **Data science angle** — MLflow helps data scientists track experiments, compare model results, package models, and manage model lifecycle metadata.
2. **MLOps angle** — MLflow creates a bridge between experimentation and production by making model development traceable and reproducible.

The main learning goal is not to memorize MLflow commands. The goal is to understand why MLflow exists and what operational problems it solves.


# 2. What is MLflow?

MLflow is an open-source platform for managing the machine learning lifecycle. In simple terms, it helps teams answer questions like:

- Which training run produced this model?
- Which parameters were used?
- Which dataset or feature set was used?
- What metrics did the model achieve?
- Where are the model artifacts stored?
- Which model version is approved for staging or production?
- Can we reproduce the result later?
- Can we roll back if a model performs badly?

Without a tool like MLflow, ML teams often rely on notebooks, manually named files, screenshots, comments, and informal memory. That may work for personal experiments, but it does not scale to production.

For a Senior SRE, the easiest analogy is this:

> MLflow gives ML teams the traceability that Git, CI/CD, artifact repositories, and deployment systems give software engineering teams.

MLflow is not the model itself. It is the metadata, artifact, packaging, and lifecycle layer around the model.


# 3. Why MLflow matters in real ML projects

A data science project is not a single script. It usually has many stages:

```text
raw data
  -> data preparation
  -> exploratory data analysis
  -> feature engineering
  -> model training
  -> model validation
  -> model packaging
  -> model registry
  -> deployment
  -> monitoring
  -> retraining
```

Every stage can change. Data can change, code can change, dependencies can change, hyperparameters can change, and evaluation metrics can change.

In normal software delivery, we would never deploy an unknown binary without knowing its source commit, build pipeline, dependency set, and test result. ML should be treated with the same discipline. MLflow helps provide that discipline for models.


# 4. The ML lifecycle and where MLflow fits

The Udemy summaries mention the lifecycle stages of a data science project: data preparation, EDA, feature engineering, model training, validation, deployment, and monitoring. Here is how MLflow fits into each stage.

| Lifecycle stage | What happens | How MLflow helps |
|---|---|---|
| Data preparation | Clean and transform source data | Track dataset references, input versions, preprocessing notes |
| EDA | Understand distributions, missing values, correlations | Log analysis artifacts, plots, summaries |
| Feature engineering | Create model inputs | Track feature logic, feature versions, transformation parameters |
| Model training | Train models with different algorithms and hyperparameters | Log parameters, metrics, artifacts, model files |
| Validation | Compare model quality | Compare runs visually and programmatically |
| Packaging | Save model in a usable format | Use MLflow Model format and flavors |
| Registry | Manage model versions | Register models, versions, aliases, lifecycle status |
| Deployment | Serve model to users or systems | Use model URI, serving patterns, deployment integration |
| Monitoring | Track post-deployment performance | Connect model version metadata to runtime metrics and drift checks |

Important: MLflow does not do all of these jobs alone. It often integrates with Git, DVC, Airflow, Kubernetes, Docker, cloud object storage, CI/CD, and observability tools.


# 5. Core components of MLflow

The lecture's visible slide lists "core components of MLflow" as the first topic. MLflow is commonly understood through four major components.

## 5.1 MLflow Tracking

Tracking records experiments and runs. A **run** is one execution of training or evaluation. A run can store:

- parameters: input configuration such as `learning_rate`, `max_depth`, `solver`, `random_state`
- metrics: outputs such as `accuracy`, `f1_score`, `rmse`, `loss`
- artifacts: files such as models, plots, reports, confusion matrices
- tags: searchable metadata such as team, owner, git commit, environment, dataset version

This is the component most learners should start with.

## 5.2 MLflow Projects

Projects provide a standard format for packaging ML code so it can be run reproducibly. From an SRE point of view, this is similar to defining a standard entrypoint for a training job.

## 5.3 MLflow Models

Models define a standard packaging format for trained models. MLflow supports multiple "flavors", for example scikit-learn, PyTorch, TensorFlow, Keras, and generic Python function models.

The goal is to reduce handoff friction between data scientists and platform engineers.

## 5.4 MLflow Model Registry

The registry manages model names, versions, lifecycle status, and approval flow. This is where production governance becomes important.

A model registry lets you say: "This exact model version is the approved candidate for staging" or "Production should use version 7."


# 6. DevOps-to-MLflow mapping

As an SRE, MLflow becomes easier when mapped to systems you already know.

| DevOps / SRE concept | MLflow concept | Why it matters |

A strong SRE position: MLflow should not be treated as a random notebook tool once it is used by teams. It should be treated as shared platform infrastructure.


# 7. Why use MLflow?

The summaries mention several reasons: experiment tracking, code structuring, pipeline creation, model packaging, dependency management, hyperparameter evaluation, visual comparison, and collaboration.

Let's expand each one.

## 7.1 Experiment tracking

Machine learning work involves experimentation. Engineers change model types, hyperparameters, feature logic, split ratios, random seeds, and datasets. Without tracking, it becomes nearly impossible to identify what produced the best result.

## 7.2 Code structuring and pipeline creation

MLflow encourages a more structured workflow. Instead of random notebook cells, the team can start thinking in terms of reproducible runs, parameters, artifacts, and versioned outputs.

## 7.3 Model packaging and dependency management

A trained model is only useful if it can be loaded and used reliably. MLflow's model format helps package model artifacts with enough metadata for downstream serving or validation.

## 7.4 Hyperparameter evaluation

Hyperparameters are configuration values that influence model behavior. MLflow makes it easier to compare multiple runs visually and programmatically.

## 7.5 Collaboration

A shared MLflow UI lets data scientists, ML engineers, and MLOps engineers inspect the same run history instead of relying on screenshots or local notebook state.


# 8. Who uses MLflow?

## 8.1 Data engineers

Data engineers care about data quality, data freshness, schema evolution, and feature availability. They may not be the primary users of MLflow, but their work affects what should be tracked.

Examples of what should be captured:

## 8.2 Data scientists

Data scientists use MLflow heavily for experiment tracking and comparison. They want to know which algorithm and parameter combination gave the best result.

## 8.3 ML engineers

ML engineers bridge data science and production engineering. They use MLflow for model packaging, validation, registry workflows, and deployment integration.

## 8.4 MLOps engineers and SREs

MLOps engineers and SREs operate the platform around MLflow. They care about:

- MLflow tracking server availability
- backend database reliability
- artifact store durability
- access control
- secrets management
- backup and restore
- network exposure
- auditability
- integration with CI/CD
- alerting and runbooks

# 11. Configure local MLflow tracking

For local learning, MLflow can use a local directory named `mlruns`.

Production should not rely on only local filesystem storage. Production setups should use a tracking server, a database backend, and durable object storage. We will keep it local here because this is the first chapter.


````python
tracking_dir = Path("mlruns").resolve()
mlflow.set_tracking_uri(tracking_dir.as_uri())
mlflow.set_experiment("lecture-53-introduction-to-mlflow")

print("Tracking URI:", mlflow.get_tracking_uri())
print("Experiment name: lecture-53-introduction-to-mlflow")
````

# 12. First MLflow run: log parameters, metrics, tags, and an artifact

This run simulates a model training result. We will log:

- parameters: configuration-like inputs
- metrics: measurable results
- tags: metadata for search and governance
- artifact: a small JSON report

This gives you the basic mental model before you train real models in later lectures.


````python
run_report = {
    "chapter": "lecture-53-introduction-to-mlflow",
    "purpose": "learn-basic-mlflow-tracking",
    "created_at": datetime.now(timezone.utc).isoformat(),
    "notes": [
        "This is a learning run.",
        "It logs parameters, metrics, tags, and an artifact.",
        "No real model is trained in this first introductory notebook."
    ]
}

artifact_path = Path("intro_run_report.json")
artifact_path.write_text(json.dumps(run_report, indent=2))

with mlflow.start_run(run_name="intro-learning-run") as run:
    mlflow.log_param("model_type", "demo-placeholder")
    mlflow.log_param("dataset", "not-applicable")
    mlflow.log_param("chapter", "introduction-to-mlflow")
    mlflow.log_param("random_seed", 42)

    mlflow.log_metric("example_accuracy", 0.91)
    mlflow.log_metric("example_f1_score", 0.89)

    mlflow.set_tag("team", "sre-mlops-learning")
    mlflow.set_tag("owner_role", "senior-sre")
    mlflow.set_tag("course_section", "complete-mlflow-tutorials")
    mlflow.set_tag("lecture", "53-introduction-to-mlflow")
    mlflow.set_tag("environment", "local-learning")

    mlflow.log_artifact(str(artifact_path), artifact_path="reports")

    print("Run ID:", run.info.run_id)
    print("Artifact logged:", artifact_path)

artifact_path.unlink(missing_ok=True)
````

# 13. Search and inspect runs programmatically

MLflow UI is useful for humans. Programmatic search is useful for automation.

In production, a CI/CD pipeline could search runs, evaluate quality gates, register the best model, or block promotion if the metrics are below threshold.


````python
experiment = mlflow.get_experiment_by_name("lecture-53-introduction-to-mlflow")
runs_df = mlflow.search_runs(experiment_ids=[experiment.experiment_id])

columns = [
    "run_id",
    "tags.mlflow.runName",
    "params.model_type",
    "params.chapter",
    "metrics.example_accuracy",
    "metrics.example_f1_score",
    "tags.owner_role",
    "status",
    "start_time",
]

runs_df[columns]
````

# 16. Production MLflow architecture

A production MLflow setup should separate the tracking server, backend store, and artifact store.

```text
Data scientist notebook / training job / CI pipeline
        |
        v
MLflow Tracking Server
        |
        +--> Backend store: PostgreSQL or MySQL
        |
        +--> Artifact store: S3 / GCS / Azure Blob / MinIO
```

## SRE production concerns

- run MLflow behind TLS
- restrict access using SSO, OAuth proxy, private ingress, or network policy
- use durable object storage for artifacts
- use managed PostgreSQL or MySQL for metadata
- back up the backend database
- test restore procedures
- monitor API latency, error rate, and availability
- alert on database failures and artifact upload/download failures
- avoid local disk as the only artifact store
- control who can approve or promote model versions


# 17. Security and governance notes

MLflow can contain valuable and sensitive information. Even if model artifacts do not contain explicit secrets, they may encode business logic, proprietary features, or sensitive behavior.

Security controls to consider:

- authentication for the MLflow UI and API
- authorization by team or project
- encryption at rest for artifact storage
- encryption in transit
- secret management for database and object store credentials
- audit logging for model registration and promotion
- retention policy for old artifacts
- data classification for artifacts and reports

Governance questions:

- Who is allowed to create experiments?
- Who owns an experiment?
- Who can register models?
- Who can promote a model to production?
- What approval is required?
- How do we roll back?
- How do we prove which model served production traffic on a given date?


# 18. MLflow and monitoring

The lecture summary mentions tracking model performance over time. This requires careful interpretation.

MLflow is excellent for tracking training-time metadata and model versions. Runtime model monitoring often requires additional systems such as Prometheus, Grafana, logs, tracing, feature drift tools, inference services, and business analytics.

A practical setup can look like this:

```text
MLflow
  -> tracks training runs, model versions, artifacts

Inference service
  -> exposes prediction API

Prometheus / Grafana
  -> tracks latency, errors, traffic, resource usage

Data drift / quality monitor
  -> tracks input distribution changes

Business metrics
  -> tracks user-impact or product-impact metrics
```

SRE takeaway: MLflow is part of the MLOps observability story, but not the entire observability stack.


# 19. Runbook starter: MLflow service degraded

## Symptoms

- MLflow UI unavailable
- training jobs fail while logging runs
- artifacts cannot be uploaded or downloaded
- model registry operations fail
- deployment pipeline cannot resolve model versions

## First checks

```bash
kubectl get pods -n mlops
kubectl get svc -n mlops
kubectl get ingress -n mlops
kubectl logs deploy/mlflow-tracking-server -n mlops --tail=100
```

## Backend checks

- Is the database reachable?
- Are credentials valid?
- Are connection limits exhausted?
- Is the database out of disk?
- Did a recent migration fail?

## Artifact store checks

- Is object storage reachable?
- Did IAM permissions change?
- Is the bucket/container policy correct?
- Are there network or DNS issues?
- Are uploads failing due to size limits?

## Mitigation

- roll back recent deployment/config changes
- restart unhealthy pods if safe
- restore previous secrets/config if rotated incorrectly
- pause model promotion if metadata integrity is uncertain
- communicate impact to ML/data teams
